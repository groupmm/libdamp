```python
import glob
import os

import libf0
import numpy as np
import soundfile as sf
```

```python
fs = 48000.0

wav_folder = "../../../datasets/VocalRevolutions/Magnificat"
ann_folder = "../../../datasets/VocalRevolutions/Magnificat"

found_files = glob.glob(os.path.join(wav_folder, "*.wav"), recursive=True)
print(len(found_files), "files found.")

os.makedirs(ann_folder, exist_ok=True)
```

```python
for wav_file in found_files:
    folder = os.path.dirname(wav_file)
    song = folder.split("/")[-2]
    basename = os.path.splitext(os.path.basename(wav_file))[0]
    f0_file = os.path.join(folder, basename + "_f0.csv")

    x, fs2 = sf.read(wav_file)
    assert fs2 == fs, "Sampling rate mismatch."

    print(x.shape)

    f0, t_f0, c_f0 = libf0.swipe(x, Fs=fs, H=256, F_min=200, F_max=1500)

    print(t_f0.shape, f0.shape, c_f0.shape)

    f0_ = np.vstack([t_f0, f0, np.nan_to_num(c_f0)]).T
    mask = f0_[:, 2] < 0.25
    f0_[mask, 1] = 0

    print(f0_.shape)

    np.savetxt(f0_file, f0_, delimiter=";", fmt="%.6f", header="t;f0;conf", comments="")
```
